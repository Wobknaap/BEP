```python
# ... (your existing code)

from umap import UMAP
from bertopic import BERTopic

# ... (your existing code for loading data and creating embeddings)

# **Stel de UMAP-parameters in voor 3D-visualisatie**
umap_n_neighbors = 40 #Aantal buren voor UMAP
umap_n_components = 3 # **Belangrijke verandering: stel in op 3 voor 3D-visualisatie**
umap_min_dist = 0.1 # Minimum afstand tussen punten in UMAP

# **Initialiseer de UMAP- en BERTopic-modellen**
# UMAP-model: Hiermee worden de embeddings gereduceerd tot 3 dimensies
umap_model = UMAP(n_neighbors=umap_n_neighbors, n_components=umap_n_components,
 min_dist=umap_min_dist, random_state=42)

# BERTopic-model: Voeg het gereduceerde UMAP-model toe
topic_model = BERTopic(umap_model=umap_model, verbose=True)

# **Train het model en verkrijg de gereduceerde embeddings**
topics, probs = topic_model.fit_transform(docs, embeddings)

# De gereduceerde 3D-embeddings zijn nu beschikbaar in umap_model.embedding_
reduced_embeddings = umap_model.embedding_
```

```python
import plotly.express as px
import pandas as pd

# Maak een DataFrame van de gereduceerde embeddings en de topics
df_plot = pd.DataFrame(reduced_embeddings, columns=['x', 'y', 'z'])
df_plot['topic'] = [str(t) for t in topics] # Zet topics om naar strings voor de plot

# Maak de interactieve 3D-scatter plot
fig = px.scatter_3d(df_plot, x='x', y='y', z='z', color='topic',
                    title='Interactieve 3D Visualisatie van BERTopic Clusters')

# Toon de plot
fig.show()
```
